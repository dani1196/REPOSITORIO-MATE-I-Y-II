# Laboratorio: Derivadas con SymPy

## Resumen
La **derivada** de $f$ en $x=a$ se define como

$$f'(a) = \lim_{h \to 0} \frac{f(a+h) - f(a)}{h}$$

Geométricamente, $f'(a)$ es la pendiente de la recta tangente a la gráfica de $f$ en el punto $(a, f(a))$.

La ecuación punto-pendiente de la recta tangente es:

$$y - y_1 = m\,(x - x_1) \quad \Longleftrightarrow \quad y =m(x-x_1) + y_1$$

Usamos $(x_1,y_1)=(a,f(a))$ y $m=f'(a)$.


```python
import sympy as sp
import numpy as np
from IPython.display import display
```

## Comandos clave de SymPy

| Comando | Descripción |
|---|---|
| `sp.Symbol('x')` | Define una variable simbólica |
| `sp.diff(f, x)` | Derivada de $f$ respecto a $x$ |
| `sp.diff(f, x, n)` | Derivada $n$-ésima |
| `f.evalf(subs={x: a})` | Evalúa numéricamente en $x=a$ |
| `sp.plot(f, (x, a, b))` | Grafica una función |

### Ejemplo

```python
# 1. Definir la variable simbólica
x = sp.Symbol('x')
# Si en lugar de x, usaramos otra variable con otra letra, por ejemplo t usaríamos
# t = sp.Symbol('t')
```

```python
# 2. Definir una función algebraica
f = x**4 - 4*x**3 + 2*x**2 + 5*x - 3
```

```python
# 3. Calcular derivadas analíticas
f_prima = sp.diff(f, x)          # Primera derivada
f_segunda = sp.diff(f, x, 2)     # Segunda derivada
f_tercera = sp.diff(f, x, 3)     # Tercera derivada
```

```python
print("\nDerivadas Superiores:")
print(f'La primera derivada es: {f_prima}')
print(f'La segunda derivada es: {f_segunda}')
print(f'La tercera derivada es: {f_tercera}')
```

```python
# Para visualizar mejor las funciones usamos el comando display
display(f_prima)
```

### Ejercicio 1
Determinar la ecuación de la recta tangente a la curva $y = \cos^2(t)$ en el punto $t = 0$.

**Pasos sugeridos:**
1. Definir $t$ como símbolo.
2. Definir $f(t) = \cos^2(t)$.
3. Calcular $f'(t)$ con `sp.diff`.
4. Evaluar $f(0)$ y $f'(0)$.
5. Construir $y = f'(0)(t-0) + f(0)$.


```python
# 1. Definir la variable simbólica
t = sp.Symbol('t')
```

```python
# 2. Definir una función algebraica
f1 = sp.cos(t)**2
```

```python
# 3. Calcular derivadas analíticas
f1_prima = sp.diff(f1, t)          # Primera derivada
print(f"y' = {f1_prima}")
```

```python
# 4. Definir el punto de evaluación
t1 = 0
y1 = f1.evalf(subs={t: t1})
m  = f1_prima.evalf(subs={t: t1})
print(f'Punto de tangencia: ({t1}, {y1:.1f})')
print(f'Pendiente: m = {m}')
```

```python
# 5. Construir la ecuación analítica de la recta tangente
recta_tangente = m*(t-t1)+y1

print(f"La ecuación de la recta tangente en t={t1} es y={recta_tangente:.1f}")
```

```python
p = sp.plot(f1, recta_tangente, (t, -sp.pi, sp.pi),
            legend=True, show=False)
p[0].line_color = 'blue'
p[1].line_color = 'red'
p.show()
```
